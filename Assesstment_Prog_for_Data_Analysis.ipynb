Research Motivation: Introduce the purpose of investigating hypertension patterns in the patient population.

1.1 Introduction

Cardiovascular diseases and other health conditions are linked to hypertension, making it a serious health problem and a public health concern. When studying hypertension, it is important to understand the factors associated with it so that the patterns can be identified at a population level and the decisions made about the management of hypertension can be better informed.

The objective of this project is to research patterns of hypertension in an anonymised dataset with 5,000 records of patient information. The data consists of demographic, anthropometric, lifestyle and clinical features, along with a target variable indicating hypertension (Yes/No).

The primary purpose of this investigation is to delve into the relationships between the various patient characteristics and health-related issues and hypertension in the dataset. Individual factors as well as relationships between two or more factors will be analyzed to find potentially meaningful relationships.

1.2 Research Questions

The following list of four research questions has been developed to guide the exploratory data analysis:

Demographic Patterns and Hypertension (Research Question 1):

What are the trends in prevalence of hypertension by age and gender? Do there seem to be particular groups of population with distinctly different prevalence patterns?

Research Question 2: What are the family history and hypertension issues of children born to parents who are diagnosed with hypertension?

Is there a difference between the prevalence of hypertension among patients with and without a family history of hypertension, and is it different in various age groups?

Research Question 3 deals with anthropometric measurements and hypertension.Research Question 3 is related to anthropometric measurements and hypertension.

What is the relationship between BMI and prevalence of hypertension, and are the association between these two measures similar or different?

Research Question 4: Lifestyle Factors and Hypertension

What is the relationship between prevalence of hypertension and lifestyle factors such as salt intake, level of physical activity, smoking status, alcohol intake, stress and sleeping time? Are there patterns when you look at combinations of these factors, that you don't see when you look at each factor individually?

1.3 Rationale for the Research Questions

RQ1 – Demographic Patterns

Demographic factors such as age and gender are key factors to study when looking at patterns of hypertension. Combining these variables into an analysis may uncover differences in hypertension prevalence among demographic groups which may not be seen in analyses conducted separately by age or gender. This investigation will determine if certain subgroups of the data set show a greater observed prevalence of hypertension.

RQ2 – Family History

Family history may be a factor in assessing the possible susceptibility to hypertension. Comparisons of patients with and without a family history will help investigate if there is any difference in hypertension prevalence between the groups. Examination of the association across age groups can provide information on the consistency of the association across the age groups.

RQ3 – Anthropometric Measurements

BMI and waist circumference are two anthropometric measures of interest to be examined with respect to hypertension. A comparison of both measurements will tell whether there is any similarity or dissimilarity in the associations with hypertension prevalence. This could give a wider range of the relationship between anthropometric features and hypertension in the data set.

RQ4 – Lifestyle Factors

Lifestyle factors can have varying relationships with hypertension. A number of behavioural risk factors can be explored: salt intake; physical activity; smoking; drinking alcohol; and lack of sleep or stress. These variables can be analyzed on an individual basis, and together to reveal patterns that may not be apparent in a comparison of only one pair of variables. This might offer some clues to the association of lifestyle factors and hypertension.



1.4 Scope of Investigation

The exploratory data analysis begins by considering these research questions. Distributions of and relationships and differences between patient groups will be explored using appropriate statistical summaries, comparisons and visualisations.

These will not be the only questions, however, which will be the basis of the analysis. Other variables such as the fasting glucose level, cholesterol and creatinine will be investigated as well, to see if there is any other trend in the data that might be important for understanding hypertension.

Special emphasis will be placed on the measurements of systolic and diastolic blood pressure. The variables in the data set may be directly related to the classification of hypertension and so will be explored carefully before deciding how to deal with them in further analysis.

The results will be discussed in relation to the data set and its limitations. Results of relationships found will be presented as associations, not as direct causation. The aim is to investigate meaningful patterns of hypertension that would help the clinical staff look into the population-level health data.

This project is aimed at exploring the demographic, family history, anthropometric and lifestyle factors that are associated with hypertension in a large dataset of 5,000 patient records. The project is aimed at discovering meaningful patterns and relationships of the data by exploratory data analysis, and then to present the patterns and relationships to clinical personnel in an interactive, educational decision-support application.
