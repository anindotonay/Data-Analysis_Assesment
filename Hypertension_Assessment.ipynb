# Task 1: Research Motivation


1.1 Introduction

Around one in three adults in the UK has high blood pressure (NHS England, 2023), and it's a key preventable risk factor for cardiovascular disease, stroke and kidney failure (NHS England, 2023). Lifestyle changes and drugs can usually control it, but it is rarely diagnosed until it causes severe complications (BHF, 2024). Identification of patients at higher risk prior to the elevation of blood pressure is therefore important and a major responsibility of primary care.

This project uses an anonymised dataset of 5,000 patients seen at a primary care clinic. It has demographic, anthropometric, lifestyle and clinical variables along with a binary target variable, indicating whether the patient has hypertension or not (Yes/No). The clinical lead wishes to have a better understanding of the factors that are associated with hypertension amongst this patient group and a device which enables staff to identify higher risk patients earlier.

The investigation thus focuses on associations between individual characteristics of patients with hypertension and combinations of these characteristics, to identify clinically relevant and clinically useful patterns.

1.2 Research Questions

The following four research questions have been developed to guide the exploratory data analysis:

Research Question 1: Demographic Patterns and Hypertension

How does hypertension prevalence vary by age and gender, and does the gender gap stay the same across age groups?

Research Question 2: Family History and Hypertension

Is hypertension more common in patients with a family history of hypertension, and does the size of this difference change across age groups?

Research Question 3: Anthropometric Measurements and Hypertension

How are BMI and waist circumference each associated with hypertension, and does waist circumference separate hypertensive from non-hypertensive patients better than BMI?

Research Question 4: Lifestyle Factors and Hypertension

How are salt intake, physical activity, smoking, alcohol consumption, stress and sleep each associated with hypertension, and does the number of modifiable risk factors a patient has show a pattern that single factors do not?

1.3 Rationale for the Research Questions

**RQ1 – Demographic Patterns**

Age and gender are the first facts a clinician knows about any patient, so they are the natural starting point for deciding who to screen first. Hypertension is generally expected to become more common with age, but it is less clear whether the difference between men and women stays the same at every age. Examining age and gender together, rather than separately, can show whether particular groups reach high prevalence earlier than others, and so whether one age threshold for screening suits everyone.

**RQ2 – Family History**

Family history is quick and free to record at a routine appointment. If patients with a family history have a higher prevalence of hypertension, it is a simple flag for increased risk. Comparing this difference across age groups shows whether the flag is equally useful for all patients, or whether it is most informative at particular ages, such as in younger patients where raised blood pressure is less often expected.

**RQ3 – Anthropometric Measurements**

BMI is the measure most often recorded in routine care, but it does not show where body fat is carried. Waist circumference reflects central fat, which is closely linked to cardiovascular risk. Because the two measures are likely to be strongly related, the question is not only whether each is associated with hypertension, but whether waist circumference separates hypertensive from non-hypertensive patients better than BMI does. The answer indicates whether routine measurement should include waist circumference.

**RQ4 – Lifestyle Factors**

Salt intake, physical activity, smoking, alcohol consumption, stress and sleep can be changed by patients, so they are the factors where advice and intervention can make a difference. Looking at each factor alone shows which are most strongly associated with hypertension. In practice, however, patients often have several risk factors at once, so counting how many modifiable risk factors a patient has may reveal a pattern that no single factor shows. This helps clinicians decide where lifestyle advice is likely to matter most.

**Exploratory questions**

The 130/80 threshold used in this dataset is stricter than the 140/90 threshold commonly used for diagnosis in UK clinics. Testing how much prevalence changes under each definition shows how far the findings depend on how hypertension is defined. Creatinine, fasting glucose and cholesterol reflect kidney and metabolic health, and may be associated with hypertension. Because all of these tend to rise with age, the association is checked within age groups so that age is not mistaken for a separate effect.

1.4 Scope of Investigation

The exploratory data analysis begins with the four research questions above. Distributions, relationships and differences between patient groups will be explored using appropriate statistical summaries, comparisons and visualisations.

The research questions are a starting point, not a limit. In addition to them, two exploratory questions will be investigated as patterns emerge from the data:

- How much does hypertension prevalence change when the dataset's 130/80 threshold is replaced by the UK clinical threshold of 140/90, and which age groups are affected most?
- Are creatinine, fasting glucose and cholesterol associated with hypertension, and does the association remain after comparing patients within the same age group?

Special emphasis will be placed on the systolic and diastolic blood pressure measurements. These variables are likely to define the hypertension label directly, so they will be examined carefully before deciding how to handle them in later analysis, to avoid circular reasoning.

The results will be discussed in relation to the dataset and its limitations. Relationships will be described as associations, not as causes, because the data is a cross-sectional snapshot. The aim is to identify meaningful patterns that could help clinical staff understand population-level health data.

1.5 Overall Research Aim

This project aims to explore the demographic, family history, anthropometric and lifestyle factors associated with hypertension in a dataset of 5,000 patient records. Through exploratory data analysis, it seeks to identify meaningful patterns and relationships, with the clinical goal of helping staff recognise higher-risk patients earlier, before their blood pressure readings escalate. These findings will then be presented to clinical staff through an interactive, educational decision-support application.
